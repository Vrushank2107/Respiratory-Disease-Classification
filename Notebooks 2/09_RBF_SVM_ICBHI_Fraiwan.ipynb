```python

import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedGroupKFold

# Merge the development manifest with the existing feature table.
# Use the manifest to establish which cycles belong to development.
join_keys = ["patient_id", "recording_id", "chest_location", "diagnosis"]

development_keys = development_manifest[join_keys].drop_duplicates()

development_features = features_df.merge(
    development_keys,
    on=join_keys,
    how="inner",
    validate="many_to_one"
)

print("Development feature rows:", len(development_features))
print("Development patients:", development_features["patient_id"].nunique())

# Try 3 folds first because the rarest diagnoses have very few patients.
for n_splits in [3, 4, 5]:
    print(f"\n=== {n_splits}-FOLD STRATIFIED GROUP CHECK ===")

    splitter = StratifiedGroupKFold(
        n_splits=n_splits,
        shuffle=True,
        random_state=42
    )

    y = development_features["diagnosis"].astype(str)
    groups = development_features["patient_id"].astype(str)

    fold_rows = []

    try:
        for fold, (train_idx, val_idx) in enumerate(
            splitter.split(development_features, y, groups),
            start=1
        ):
            train_df = development_features.iloc[train_idx]
            val_df = development_features.iloc[val_idx]

            train_classes = set(train_df["diagnosis"])
            val_classes = set(val_df["diagnosis"])

            fold_rows.append({
                "fold": fold,
                "train_patients": train_df["patient_id"].nunique(),
                "validation_patients": val_df["patient_id"].nunique(),
                "train_classes": len(train_classes),
                "validation_classes": len(val_classes),
                "missing_validation_classes": ", ".join(
                    sorted(set(y.unique()) - val_classes)
                ) or "None",
                "patient_overlap": len(
                    set(train_df["patient_id"].astype(str))
                    & set(val_df["patient_id"].astype(str))
                )
            })

        display(pd.DataFrame(fold_rows))

    except ValueError as exc:
        print("Fold generation failed:", exc)

print("\nNo models were trained and no files were modified.")


```

```python

# Determine whether duplicate keys are expected multiple cycles or
# whether the manifest-to-feature merge multiplied rows.

key_columns = [
    "patient_id",
    "recording_id",
    "chest_location",
    "diagnosis",
]

print("=== ROW COUNTS ===")
print("Full feature table:", len(features_df))
print("Development manifest:", len(development_manifest))
print("Merged development features:", len(development_features))

print("\n=== KEY UNIQUENESS ===")
print(
    "Unique keys in development manifest:",
    development_manifest[key_columns].drop_duplicates().shape[0]
)
print(
    "Unique keys in merged development features:",
    development_features[key_columns].drop_duplicates().shape[0]
)

print("\n=== FEATURE ROWS PER KEY ===")
feature_key_counts = (
    development_features.groupby(key_columns, dropna=False)
    .size()
    .rename("rows_per_key")
    .reset_index()
)

print("Keys with multiple feature rows:", int(
    (feature_key_counts["rows_per_key"] > 1).sum()
))
print("Maximum rows for one key:", int(feature_key_counts["rows_per_key"].max()))

print("\nExamples of keys with multiple rows:")
display(
    feature_key_counts[
        feature_key_counts["rows_per_key"] > 1
    ].head(10)
)

print("\n=== ORIGINAL FEATURE TABLE DUPLICATE KEY COUNTS ===")
original_dev = features_df.merge(
    development_manifest[key_columns].drop_duplicates(),
    on=key_columns,
    how="inner",
    validate="many_to_one"
)

print("Original feature rows matching development keys:", len(original_dev))
print(
    "Duplicate full rows in matched feature data:",
    int(original_dev.duplicated().sum())
)

print("\nNo files were modified.")
```

```python

from sklearn.base import clone
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    classification_report,
)

# Reproducible baseline; no hyperparameter search yet.
baseline_model = Pipeline([
    ("scaler", StandardScaler()),
    ("svm", SVC(
        kernel="rbf",
        class_weight="balanced",
        C=1.0,
        gamma="scale",
        probability=False,
        random_state=42,
    )),
])

cv = StratifiedGroupKFold(
    n_splits=3,
    shuffle=True,
    random_state=42,
)

baseline_results = []
baseline_oof = pd.DataFrame({
    "patient_id": groups_dev.to_numpy(),
    "y_true": y_dev.to_numpy(),
    "y_pred": [None] * len(y_dev),
})

for fold, (train_idx, val_idx) in enumerate(
    cv.split(X_dev, y_dev, groups_dev),
    start=1,
):
    model = clone(baseline_model)

    X_train, X_val = X_dev.iloc[train_idx], X_dev.iloc[val_idx]
    y_train, y_val = y_dev.iloc[train_idx], y_dev.iloc[val_idx]

    model.fit(X_train, y_train)
    predictions = model.predict(X_val)

    baseline_oof.loc[val_idx, "y_pred"] = predictions

    baseline_results.append({
        "fold": fold,
        "train_patients": groups_dev.iloc[train_idx].nunique(),
        "validation_patients": groups_dev.iloc[val_idx].nunique(),
        "accuracy": accuracy_score(y_val, predictions),
        "balanced_accuracy": balanced_accuracy_score(y_val, predictions),
        "macro_f1": f1_score(
            y_val, predictions, average="macro", zero_division=0
        ),
        "weighted_f1": f1_score(
            y_val, predictions, average="weighted", zero_division=0
        ),
    })

baseline_results_df = pd.DataFrame(baseline_results)

print("=== PATIENT-GROUPED RBF-SVM BASELINE ===")
display(baseline_results_df.round(4))

print("\n=== FOLD MEAN AND STANDARD DEVIATION ===")
display(
    baseline_results_df[
        ["accuracy", "balanced_accuracy", "macro_f1", "weighted_f1"]
    ].agg(["mean", "std"]).round(4)
)

print("\n=== OUT-OF-FOLD CLASSIFICATION REPORT ===")
print(classification_report(
    baseline_oof["y_true"],
    baseline_oof["y_pred"],
    labels=sorted(y_dev.unique()),
    zero_division=0,
))

print("No model files were saved or overwritten.")
```

```python

from sklearn.feature_selection import f_classif, mutual_info_classif
from sklearn.ensemble import RandomForestClassifier

def select_features_within_fold(
    X_train_fold,
    y_train_fold,
    correlation_threshold=0.90,
    random_state=42,
):
    """
    Reproduce Notebook 07's consensus feature selection using
    only the current training fold.
    """
    X = X_train_fold[candidate_features].copy()
    y = pd.Series(y_train_fold, index=X.index)

    # 1. ANOVA scores
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        anova_scores, _ = f_classif(X, y)

    # 2. Mutual-information scores
    mi_scores = mutual_info_classif(
        X,
        y,
        random_state=random_state,
    )

    # 3. Random Forest importance
    rf = RandomForestClassifier(
        n_estimators=300,
        random_state=random_state,
        n_jobs=-1,
        class_weight="balanced",
    )
    rf.fit(X, y)
    rf_scores = rf.feature_importances_

    score_table = pd.DataFrame({
        "feature": candidate_features,
        "anova": np.nan_to_num(
            anova_scores, nan=0.0, posinf=0.0, neginf=0.0
        ),
        "mutual_information": np.nan_to_num(
            mi_scores, nan=0.0, posinf=0.0, neginf=0.0
        ),
        "random_forest": np.nan_to_num(
            rf_scores, nan=0.0, posinf=0.0, neginf=0.0
        ),
    })

    # 4. Rank descending, then combine normalized ranks
    for score_name in ["anova", "mutual_information", "random_forest"]:
        rank_name = f"{score_name}_rank"
        normalized_name = f"{score_name}_normalized"

        score_table[rank_name] = score_table[score_name].rank(
            ascending=False,
            method="min",
        )

        max_rank = score_table[rank_name].max()
        score_table[normalized_name] = (
            max_rank - score_table[rank_name] + 1
        ) / max_rank

    normalized_columns = [
        "anova_normalized",
        "mutual_information_normalized",
        "random_forest_normalized",
    ]

    score_table["consensus_score"] = score_table[
        normalized_columns
    ].mean(axis=1)

    score_table = score_table.sort_values(
        "consensus_score",
        ascending=False,
        kind="stable",
    ).reset_index(drop=True)

    # 5. Greedy redundancy filtering using training-fold correlations
    corr = X.corr().abs().fillna(0.0)
    selected = []

    for feature in score_table["feature"]:
        if not selected:
            selected.append(feature)
            continue

        max_correlation = corr.loc[feature, selected].max()

        if max_correlation < correlation_threshold:
            selected.append(feature)

    if not selected:
        raise ValueError("Feature selection returned no features.")

    return selected, score_table

```

```python

from sklearn.model_selection import StratifiedGroupKFold

cv = StratifiedGroupKFold(
    n_splits=3,
    shuffle=True,
    random_state=42,
)

# Get the first patient-grouped fold
train_idx, val_idx = next(
    cv.split(X_dev, y_dev, groups=groups_dev)
)

X_train_fold = X_dev.iloc[train_idx]
y_train_fold = y_dev.iloc[train_idx]
groups_train_fold = groups_dev.iloc[train_idx]

X_val_fold = X_dev.iloc[val_idx]
y_val_fold = y_dev.iloc[val_idx]
groups_val_fold = groups_dev.iloc[val_idx]

# Confirm there is no patient overlap
patient_overlap = set(groups_train_fold) & set(groups_val_fold)
assert len(patient_overlap) == 0, "Patient leakage detected!"

# Select features using only the training fold
fold_features, fold_ranking = select_features_within_fold(
    X_train_fold,
    y_train_fold,
    correlation_threshold=0.90,
    random_state=42,
)

print("Training cycles:", len(train_idx))
print("Validation cycles:", len(val_idx))
print("Training patients:", groups_train_fold.nunique())
print("Validation patients:", groups_val_fold.nunique())
print("Patient overlap:", len(patient_overlap))

print("\nSelected features:", len(fold_features))
print("Selected feature names:")
for i, feature in enumerate(fold_features, start=1):
    print(f"{i:02d}. {feature}")

print("\nTop 10 consensus-ranked features:")
display(
    fold_ranking[
        ["feature", "anova", "mutual_information",
         "random_forest", "consensus_score"]
    ].head(10).round(4)
)

assert len(fold_features) > 0
assert set(fold_features).issubset(set(candidate_features))

print("\nPASS: fold-specific feature selection completed.")
```

```python

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix,
)

fold_model = Pipeline([
    ("scaler", StandardScaler()),
    ("svm", SVC(
        kernel="rbf",
        C=1.0,
        gamma="scale",
        class_weight="balanced",
        probability=False,
        random_state=42,
    )),
])

# Train only on this fold's training patients and selected features
fold_model.fit(
    X_train_fold[fold_features],
    y_train_fold,
)

# Predict the unseen validation patients
fold_predictions = fold_model.predict(
    X_val_fold[fold_features]
)

print("Patient overlap:", len(set(groups_train_fold) & set(groups_val_fold)))
print("Features used:", len(fold_features))

print("\nFOLD 1 METRICS")
print("Accuracy:", round(accuracy_score(y_val_fold, fold_predictions), 4))
print(
    "Balanced accuracy:",
    round(balanced_accuracy_score(y_val_fold, fold_predictions), 4),
)
print(
    "Macro-F1:",
    round(
        f1_score(
            y_val_fold,
            fold_predictions,
            labels=ALL_CLASSES,
            average="macro",
            zero_division=0,
        ),
        4,
    ),
)
print(
    "Weighted-F1:",
    round(
        f1_score(
            y_val_fold,
            fold_predictions,
            average="weighted",
            zero_division=0,
        ),
        4,
    ),
)

print("\nClassification report:")
print(
    classification_report(
        y_val_fold,
        fold_predictions,
        labels=ALL_CLASSES,
        zero_division=0,
    ),
)

print("Confusion matrix (rows=true, columns=predicted):")
display(
    pd.DataFrame(
        confusion_matrix(
            y_val_fold,
            fold_predictions,
            labels=ALL_CLASSES,
        ),
        index=ALL_CLASSES,
        columns=ALL_CLASSES,
    )
)
```

```python

from sklearn.base import clone
from sklearn.metrics import confusion_matrix

# Fixed, patient-grouped splits
fold_splits = list(cv.split(X_dev, y_dev, groups=groups_dev))

oof_predictions = np.empty(len(y_dev), dtype=object)
oof_predictions[:] = None
fold_results = []
fold_feature_counts = []

for fold_number, (train_idx, val_idx) in enumerate(fold_splits, start=1):
    X_train = X_dev.iloc[train_idx]
    y_train = y_dev.iloc[train_idx]
    groups_train = groups_dev.iloc[train_idx]

    X_val = X_dev.iloc[val_idx]
    y_val = y_dev.iloc[val_idx]
    groups_val = groups_dev.iloc[val_idx]

    # Confirm patient-level separation
    overlap = set(groups_train) & set(groups_val)
    assert not overlap, f"Patient leakage in fold {fold_number}"

    # Select features using only the training portion
    fold_features, ranking = select_features_within_fold(
        X_train,
        y_train,
        correlation_threshold=0.90,
        random_state=42,
    )

    model = Pipeline([
        ("scaler", StandardScaler()),
        ("svm", SVC(
            kernel="rbf",
            C=1.0,
            gamma="scale",
            class_weight="balanced",
            probability=False,
            random_state=42,
        )),
    ])

    model.fit(X_train[fold_features], y_train)
    predictions = model.predict(X_val[fold_features])

    # Store predictions at their original row positions
    oof_predictions[val_idx] = predictions
    fold_feature_counts.append(len(fold_features))

    fold_results.append({
        "fold": fold_number,
        "train_cycles": len(train_idx),
        "validation_cycles": len(val_idx),
        "train_patients": groups_train.nunique(),
        "validation_patients": groups_val.nunique(),
        "selected_features": len(fold_features),
        "accuracy": accuracy_score(y_val, predictions),
        "balanced_accuracy": balanced_accuracy_score(y_val, predictions),
        "macro_f1": f1_score(
            y_val, predictions, labels=ALL_CLASSES,
            average="macro", zero_division=0
        ),
        "weighted_f1": f1_score(
            y_val, predictions, average="weighted", zero_division=0
        ),
    })

    print(
        f"Fold {fold_number}/3 complete | "
        f"features={len(fold_features)} | "
        f"accuracy={accuracy_score(y_val, predictions):.4f} | "
        f"macro-F1={fold_results[-1]['macro_f1']:.4f}"
    )

assert all(pred is not None for pred in oof_predictions)

fold_results_df = pd.DataFrame(fold_results)

print("\nPER-FOLD RESULTS")
display(fold_results_df.round(4))

print("\nMEAN AND STANDARD DEVIATION ACROSS FOLDS")
display(
    fold_results_df[
        ["accuracy", "balanced_accuracy", "macro_f1", "weighted_f1"]
    ].agg(["mean", "std"]).round(4)
)

print("\nPOOLED OUT-OF-FOLD RESULTS")
print("Accuracy:", round(accuracy_score(y_dev, oof_predictions), 4))
print(
    "Balanced accuracy:",
    round(balanced_accuracy_score(y_dev, oof_predictions), 4),
)
print(
    "Macro-F1:",
    round(f1_score(
        y_dev, oof_predictions, labels=ALL_CLASSES,
        average="macro", zero_division=0
    ), 4),
)
print(
    "Weighted-F1:",
    round(f1_score(
        y_dev, oof_predictions, average="weighted", zero_division=0
    ), 4),
)

print("\nPOOLED CLASSIFICATION REPORT")
print(classification_report(
    y_dev,
    oof_predictions,
    labels=ALL_CLASSES,
    zero_division=0,
))

print("Selected feature counts by fold:", fold_feature_counts)
```
