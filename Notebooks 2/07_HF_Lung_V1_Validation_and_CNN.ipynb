```python

from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_ROOT = Path.cwd().parent
HF_ROOT = PROJECT_ROOT / "data" / "raw" / "HF Lung V1" / "HF_Lung_V1-master"
DATASET_ROOT = HF_ROOT / "Dataset"
MANIFEST_PATH = (
    PROJECT_ROOT / "data" / "processed" / "unified_phenotype"
    / "hf_lung_v1" / "hf_phenotype_manifest.csv"
)

manifest = pd.read_csv(MANIFEST_PATH)

# Raw audio and label file inventories.
raw_audio = list(DATASET_ROOT.rglob("*.wav"))
raw_labels = list(DATASET_ROOT.rglob("*_label.txt"))

audio_stems = {p.stem for p in raw_audio}
label_stems = {p.stem.removesuffix("_label") for p in raw_labels}

print("=== RAW FILE COVERAGE ===")
print("Raw WAV files:", len(raw_audio))
print("Raw label files:", len(raw_labels))
print("Audio without matching label:", len(audio_stems - label_stems))
print("Label without matching audio:", len(label_stems - audio_stems))

print("\n=== MANIFEST RECORDING COVERAGE ===")
print("Manifest rows (phases):", len(manifest))
print("Unique recording IDs:", manifest["recording_id"].nunique())
print("Unique source audio paths:", manifest["source_audio_path"].nunique())
print("Unique annotation paths:", manifest["annotation_path"].nunique())

print("\nRows per split:")
print(manifest.groupby("split").size().to_string())

print("\nUnique recordings per split:")
print(
    manifest.groupby("split")["recording_id"]
    .nunique()
    .to_string()
)

print("\n=== SPLIT LEAKAGE CHECK ===")
recordings_by_split = {
    split: set(group["recording_id"].dropna().unique())
    for split, group in manifest.groupby("split")
}

split_names = sorted(recordings_by_split)
for i, split_a in enumerate(split_names):
    for split_b in split_names[i + 1:]:
        overlap = recordings_by_split[split_a] & recordings_by_split[split_b]
        print(f"{split_a} vs {split_b}: {len(overlap)} overlapping recording IDs")

# Same-date split leakage: a date should not occur in multiple partitions.
date_split_counts = (
    manifest.dropna(subset=["recording_date", "split"])
    .groupby("recording_date")["split"]
    .nunique()
)
leaked_dates = date_split_counts[date_split_counts > 1]

print("\n=== DATE-LEVEL SPLIT CHECK ===")
print("Unique dates:", manifest["recording_date"].nunique())
print("Dates appearing in multiple splits:", len(leaked_dates))
if len(leaked_dates):
    print(leaked_dates.head(20).to_string())

# Validate intervals and duration consistency.
start = pd.to_numeric(manifest["start_time"], errors="coerce")
end = pd.to_numeric(manifest["end_time"], errors="coerce")
duration = pd.to_numeric(manifest["duration_seconds"], errors="coerce")

invalid_intervals = (
    start.isna() | end.isna() | duration.isna()
    | (start < 0) | (end <= start)
    | ~np.isclose(end - start, duration, atol=0.01, rtol=0)
)

print("\n=== PHASE INTERVAL CHECK ===")
print("Rows with missing/non-numeric interval values:", int((start.isna() | end.isna() | duration.isna()).sum()))
print("Rows with invalid intervals or duration mismatch:", int(invalid_intervals.sum()))
print("Non-15-second source recordings (from manifest phase endpoints > 15s):",
      int((end > 15.01).sum()))

print("\n=== LABEL AND SAMPLING-RATE CHECK ===")
print("Phase types:")
print(manifest["phase_type"].value_counts(dropna=False).to_string())

print("\nPhenotype labels:")
print(manifest["phenotype_label"].value_counts(dropna=False).to_string())

print("\nSample rates:")
print(manifest["sample_rate"].value_counts(dropna=False).to_string())

print("\nAll checks completed. No files were modified.")
```

```python

from pathlib import Path
from collections import Counter
import pandas as pd

PROJECT_ROOT = Path.cwd().parent
HF_ROOT = PROJECT_ROOT / "data" / "raw" / "HF Lung V1" / "HF_Lung_V1-master"
DATASET_ROOT = HF_ROOT / "Dataset"
MANIFEST_PATH = (
    PROJECT_ROOT / "data" / "processed" / "unified_phenotype"
    / "hf_lung_v1" / "hf_phenotype_manifest.csv"
)

manifest = pd.read_csv(MANIFEST_PATH)

raw_audio = {p.stem: p for p in DATASET_ROOT.rglob("*.wav")}
raw_labels = {
    p.stem.removesuffix("_label"): p
    for p in DATASET_ROOT.rglob("*_label.txt")
}
manifest_ids = set(manifest["recording_id"].dropna().astype(str))
missing_ids = sorted(set(raw_audio) - manifest_ids)

def read_codes(label_path):
    codes = []
    for line in label_path.read_text(errors="replace").splitlines():
        parts = line.split()
        if parts:
            codes.append(parts[0])
    return codes

# Summarize raw annotation patterns for recordings that were excluded.
pattern_counts = Counter()
missing_summary = []

for rid in missing_ids:
    codes = read_codes(raw_labels[rid])
    pattern = tuple(sorted(set(codes)))
    pattern_counts[pattern] += 1
    missing_summary.append({
        "recording_id": rid,
        "raw_split": raw_audio[rid].parent.name,
        "annotation_lines": len(codes),
        "codes": ",".join(pattern) if pattern else "(empty)",
    })

print("=== Most common annotation-code patterns among missing recordings ===")
for pattern, count in pattern_counts.most_common(25):
    print(f"{pattern}: {count}")

print("\n=== Raw split counts among missing recordings ===")
print(pd.DataFrame(missing_summary)["raw_split"].value_counts().to_string())

# Compare manifest phase-level annotation-derived labels.
print("\n=== Manifest phenotype by phase type ===")
print(pd.crosstab(
    manifest["phase_type"],
    manifest["phenotype_label"],
    margins=True
).to_string())

print("\n=== Manifest label interpretation ===")
print(
    manifest.groupby(["phenotype_label", "label_interpretation"])
    .size()
    .sort_values(ascending=False)
    .to_string()
)

print("\n=== Missing recording examples by annotation pattern ===")
summary_df = pd.DataFrame(missing_summary)
display(summary_df.head(20))

# Inspect a few annotation files with different patterns.
print("\n=== Raw annotation excerpts ===")
shown = 0
for row in missing_summary:
    codes = row["codes"]
    if codes != "(empty)" and shown < 5:
        rid = row["recording_id"]
        print(f"\n--- {rid} | codes={codes} ---")
        print(raw_labels[rid].read_text(errors="replace")[:1200])
        shown += 1
```

```python

from pathlib import Path
from collections import Counter
import pandas as pd
import numpy as np
import re

PROJECT_ROOT = Path.cwd().parent
HF_ROOT = PROJECT_ROOT / "data" / "raw" / "HF Lung V1" / "HF_Lung_V1-master"
DATASET_ROOT = HF_ROOT / "Dataset"

audio_paths = sorted(DATASET_ROOT.rglob("*.wav"))
label_paths = {
    p.stem.removesuffix("_label"): p
    for p in DATASET_ROOT.rglob("*_label.txt")
}

# Annotation format: CODE HH:MM:SS.mmm HH:MM:SS.mmm
time_pattern = re.compile(r"^\d{2}:\d{2}:\d{2}(?:\.\d+)?$")

def parse_timestamp(value):
    hours, minutes, seconds = value.split(":")
    return int(hours) * 3600 + int(minutes) * 60 + float(seconds)

valid_codes = {"I", "E", "D", "Wheeze", "Rhonchi", "Stridor"}
records = []
interval_errors = []

for audio_path in audio_paths:
    recording_id = audio_path.stem
    label_path = label_paths.get(recording_id)

    if label_path is None:
        records.append({
            "recording_id": recording_id,
            "split": audio_path.parent.name,
            "audio_path": str(audio_path),
            "annotation_path": None,
            "annotation_lines": 0,
            "codes": (),
            "empty_annotation": True,
            "invalid_lines": 0,
            "invalid_intervals": 0,
            "event_counts": {},
        })
        continue

    lines = [
        line.strip()
        for line in label_path.read_text(errors="replace").splitlines()
        if line.strip()
    ]

    code_counts = Counter()
    invalid_lines = 0
    invalid_intervals = 0
    codes = set()

    for line_number, line in enumerate(lines, start=1):
        parts = line.split()

        if len(parts) != 3 or parts[0] not in valid_codes:
            invalid_lines += 1
            interval_errors.append({
                "recording_id": recording_id,
                "line_number": line_number,
                "issue": "Malformed line or unknown code",
                "line": line,
            })
            continue

        code, start_text, end_text = parts

        if not time_pattern.match(start_text) or not time_pattern.match(end_text):
            invalid_lines += 1
            interval_errors.append({
                "recording_id": recording_id,
                "line_number": line_number,
                "issue": "Invalid timestamp format",
                "line": line,
            })
            continue

        start = parse_timestamp(start_text)
        end = parse_timestamp(end_text)

        if start < 0 or end <= start or end > 15.01:
            invalid_intervals += 1
            interval_errors.append({
                "recording_id": recording_id,
                "line_number": line_number,
                "issue": "Invalid interval bounds",
                "line": line,
            })
            continue

        codes.add(code)
        code_counts[code] += 1

    records.append({
        "recording_id": recording_id,
        "split": audio_path.parent.name,
        "audio_path": str(audio_path),
        "annotation_path": str(label_path),
        "annotation_lines": len(lines),
        "codes": tuple(sorted(codes)),
        "empty_annotation": len(lines) == 0,
        "invalid_lines": invalid_lines,
        "invalid_intervals": invalid_intervals,
        "event_counts": dict(code_counts),
    })

raw_manifest = pd.DataFrame(records)

print("=== RECORDING-LEVEL AUDIT ===")
print("Total raw recordings:", len(raw_manifest))
print("Unique recording IDs:", raw_manifest["recording_id"].nunique())
print("Duplicate recording IDs:", int(raw_manifest["recording_id"].duplicated().sum()))
print("\nRecordings by split:")
print(raw_manifest.groupby("split").size().to_string())

print("\nEmpty annotation files:", int(raw_manifest["empty_annotation"].sum()))
print("Recordings with malformed annotation lines:",
      int((raw_manifest["invalid_lines"] > 0).sum()))
print("Recordings with invalid intervals:",
      int((raw_manifest["invalid_intervals"] > 0).sum()))
print("Total malformed lines:", int(raw_manifest["invalid_lines"].sum()))
print("Total invalid intervals:", int(raw_manifest["invalid_intervals"].sum()))

print("\n=== RECORDINGS WITH EACH ANNOTATION CODE ===")
for code in sorted(valid_codes):
    count = raw_manifest["codes"].apply(lambda codes: code in codes).sum()
    print(f"{code}: {count}")

print("\n=== ANNOTATION LINE COUNTS BY SPLIT ===")
print(
    raw_manifest.groupby("split")["annotation_lines"]
    .agg(["count", "sum", "mean", "min", "max"])
    .to_string()
)

print("\n=== ERROR EXAMPLES ===")
display(pd.DataFrame(interval_errors).head(20))

print("\nAudit complete. No files were written or modified.")
```

```python

from pathlib import Path
from collections import Counter
import soundfile as sf
import pandas as pd
import numpy as np

PROJECT_ROOT = Path.cwd().parent
HF_ROOT = PROJECT_ROOT / "data" / "raw" / "HF Lung V1" / "HF_Lung_V1-master"
DATASET_ROOT = HF_ROOT / "Dataset"

audio_paths = sorted(DATASET_ROOT.rglob("*.wav"))
audio_results = []

for i, path in enumerate(audio_paths, start=1):
    try:
        info = sf.info(str(path))
        duration = info.frames / info.samplerate if info.samplerate else np.nan

        audio_results.append({
            "recording_id": path.stem,
            "split": path.parent.name,
            "path": str(path),
            "sample_rate": info.samplerate,
            "channels": info.channels,
            "frames": info.frames,
            "duration_seconds": duration,
            "subtype": info.subtype,
            "error": None,
        })
    except Exception as exc:
        audio_results.append({
            "recording_id": path.stem,
            "split": path.parent.name,
            "path": str(path),
            "sample_rate": np.nan,
            "channels": np.nan,
            "frames": np.nan,
            "duration_seconds": np.nan,
            "subtype": None,
            "error": repr(exc),
        })

audio_audit = pd.DataFrame(audio_results)

print("=== AUDIO HEADER AUDIT ===")
print("Files inspected:", len(audio_audit))
print("Unreadable/header errors:", int(audio_audit["error"].notna().sum()))

print("\nSample-rate distribution:")
print(audio_audit["sample_rate"].value_counts(dropna=False).sort_index().to_string())

print("\nChannel-count distribution:")
print(audio_audit["channels"].value_counts(dropna=False).sort_index().to_string())

print("\nDuration summary (seconds):")
print(audio_audit["duration_seconds"].describe().to_string())

print("\nDuration distribution, rounded to 3 decimals:")
print(
    audio_audit["duration_seconds"]
    .round(3)
    .value_counts()
    .sort_index()
    .to_string()
)

bad_duration = (
    audio_audit["duration_seconds"].isna()
    | (audio_audit["duration_seconds"] <= 0)
    | (audio_audit["duration_seconds"] > 15.05)
    | (audio_audit["duration_seconds"] < 14.95)
)

print("\nRecordings outside 14.95–15.05 seconds:", int(bad_duration.sum()))
print("\nExamples outside expected duration:")
display(
    audio_audit.loc[
        bad_duration,
        ["recording_id", "split", "sample_rate", "channels",
         "duration_seconds", "error"]
    ].head(30)
)

print("\nAudit complete. Audio files were not modified.")
```

```python

from pathlib import Path
from collections import defaultdict
import hashlib
import pandas as pd

PROJECT_ROOT = Path.cwd().parent
HF_ROOT = PROJECT_ROOT / "data" / "raw" / "HF Lung V1" / "HF_Lung_V1-master"
DATASET_ROOT = HF_ROOT / "Dataset"

audio_paths = sorted(DATASET_ROOT.rglob("*.wav"))

def sha256_file(path, chunk_size=1024 * 1024):
    digest = hashlib.sha256()
    with open(path, "rb") as f:
        while True:
            chunk = f.read(chunk_size)
            if not chunk:
                break
            digest.update(chunk)
    return digest.hexdigest()

hash_records = []

for i, path in enumerate(audio_paths, start=1):
    hash_records.append({
        "recording_id": path.stem,
        "split": path.parent.name,
        "path": str(path),
        "sha256": sha256_file(path),
    })

hash_df = pd.DataFrame(hash_records)

print("Files hashed:", len(hash_df))
print("Unique SHA-256 hashes:", hash_df["sha256"].nunique())
print("Duplicate files beyond first copy:",
      len(hash_df) - hash_df["sha256"].nunique())

duplicate_groups = (
    hash_df.groupby("sha256")
    .filter(lambda group: len(group) > 1)
    .sort_values(["sha256", "split", "recording_id"])
)

print("\n=== EXACT DUPLICATE GROUPS ===")
if duplicate_groups.empty:
    print("No exact duplicate WAV files found.")
else:
    display(duplicate_groups[["recording_id", "split", "sha256", "path"]])

    print("\nDuplicate group count:",
          duplicate_groups["sha256"].nunique())

    cross_split_groups = (
        duplicate_groups.groupby("sha256")["split"]
        .nunique()
        .gt(1)
    )
    print("Duplicate groups spanning multiple splits:",
          int(cross_split_groups.sum()))

print("\nNo files were modified.")
```

```python

from pathlib import Path
import hashlib
import pandas as pd

PROJECT_ROOT = Path.cwd().parent

HF_ROOT = (
    PROJECT_ROOT / "data" / "raw" /
    "HF Lung V1" / "HF_Lung_V1-master" / "Dataset"
)

# Search for audio files in the existing project data directories.
# Exclude HF here because it is scanned separately below.
search_roots = [
    PROJECT_ROOT / "data" / "raw",
]

def sha256_file(path, chunk_size=1024 * 1024):
    digest = hashlib.sha256()
    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(chunk_size), b""):
            digest.update(chunk)
    return digest.hexdigest()

# Locate WAVs outside the HF Lung V1 directory.
other_audio = []
hf_resolved = HF_ROOT.resolve()

for root in search_roots:
    if not root.exists():
        continue
    for path in root.rglob("*"):
        if not path.is_file() or path.suffix.lower() != ".wav":
            continue
        try:
            path.resolve().relative_to(hf_resolved)
            continue
        except ValueError:
            pass
        other_audio.append(path)

hf_audio = sorted(HF_ROOT.rglob("*.wav"))
print("HF WAV files:", len(hf_audio))
print("Other raw WAV files found:", len(other_audio))

# Hash HF files and all other raw WAV files.
hf_hashes = {}
for path in hf_audio:
    hf_hashes.setdefault(sha256_file(path), []).append(str(path))

other_hashes = {}
for path in other_audio:
    other_hashes.setdefault(sha256_file(path), []).append(str(path))

shared_hashes = set(hf_hashes) & set(other_hashes)

print("\n=== CROSS-DATASET EXACT AUDIO OVERLAP ===")
print("Shared exact hashes:", len(shared_hashes))

if shared_hashes:
    overlap_rows = []
    for digest in sorted(shared_hashes):
        for hf_path in hf_hashes[digest]:
            for other_path in other_hashes[digest]:
                overlap_rows.append({
                    "sha256": digest,
                    "hf_path": hf_path,
                    "other_path": other_path,
                })
    overlap_df = pd.DataFrame(overlap_rows)
    display(overlap_df)
else:
    print("No exact HF audio matches found in other raw WAV files.")

print("\nRead-only audit complete.")
```
